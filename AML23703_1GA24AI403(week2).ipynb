```python
#4
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector
from qiskit.visualization import plot_bloch_multivector
import matplotlib.pyplot as plt
import numpy as np
import random
```

```python
qc = QuantumCircuit(1)
```

```python
qc.h(0)
```

```python
theta = random.uniform(-0.2, 0.2)

print("Random Phase Error (radians):", theta)
```

```python
qc.p(theta, 0)
```

```python
plot_bloch_multivector(state)
plt.show()
```

```python
#5
from qiskit.quantum_info import Statevector
from qiskit.visualization import plot_bloch_multivector
import matplotlib.pyplot as plt
import numpy as np

def create_qubit(theta, phi):
    """
    Construct an arbitrary single-qubit state
    |ψ> = cos(theta/2)|0> + e^(iφ) sin(theta/2)|1>
    """

   
    alpha = np.cos(theta / 2)
    beta = np.exp(1j * phi) * np.sin(theta / 2)

    
    state = Statevector([alpha, beta])

    print("Theta =", theta)
    print("Phi =", phi)

    print("\nStatevector:")
    print(state)

    print("\nManual amplitudes:")
    print("alpha =", alpha)
    print("beta  =", beta)

   
    if np.allclose(state.data, [alpha, beta]):
        print("\nValidation Successful!")
    else:
        print("\nValidation Failed!")

    
    plot_bloch_multivector(state)
    plt.show()

    return state



theta = np.pi / 3      
phi = np.pi / 4       

create_qubit(theta, phi)
```
