```python
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator
oracle = QuantumCircuit(2)
qc = QuantumCircuit(2, 1)
qc.x(1)
qc.h(0)
qc.h(1)
qc.compose(oracle, inplace=True)
qc.h(0)
qc.measure(0, 0)
print(qc.draw(output="text"))
simulator = AerSimulator()

result = simulator.run(
    qc,
    shots=1024
).result()

counts = result.get_counts()

print("Constant oracle:", counts)
```

```python
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator

oracle = QuantumCircuit(2)
oracle.cx(0, 1)
qc = QuantumCircuit(2, 1)
qc.x(1)
qc.h(0)
qc.h(1)

qc.compose(oracle, inplace=True)
qc.h(0)
qc.measure(0, 0)
print(qc.draw())

simulator = AerSimulator()

result = simulator.run(
    qc,
    shots=1024
).result()

counts = result.get_counts()

print("Balanced oracle:", counts)
```

```python
from qiskit import QuantumCircuit

def create_oracle(function_type):
    oracle = QuantumCircuit(2)

    if function_type == "constant":
        # f(x) = 0
        pass

    elif function_type == "balanced":
        # f(x) = x
        oracle.cx(0, 1)

    else:
        raise ValueError("Choose 'constant' or 'balanced'")

    return oracle
```

```python
constant_oracle = create_oracle("constant")
balanced_oracle = create_oracle("balanced")

print("Constant oracle:")
print(constant_oracle)

print("Balanced oracle:")
print(balanced_oracle)
```

```python
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator
import numpy as np

def run_experiment(error_angle, shots=1000):

    qc = QuantumCircuit(2, 1)
    qc.x(1)
    qc.h(0)
    qc.h(1)
    qc.cx(0, 1)
    qc.h(0)
    qc.ry(error_angle, 0)
    qc.measure(0, 0)

    simulator = AerSimulator()

    result = simulator.run(
        qc,
        shots=shots
    ).result()

    return result.get_counts()
```

```python
errors = [0, 0.05, 0.10, 0.20, 0.30]

for error in errors:
    counts = run_experiment(error, shots=1000)

    print("Error angle:", error)
    print("Measurement:", counts)
    print()
```
