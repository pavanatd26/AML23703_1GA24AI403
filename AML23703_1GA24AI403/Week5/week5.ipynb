```python
import numpy as np

from qiskit import QuantumCircuit, transpile
from qiskit_aer import AerSimulator

simulator = AerSimulator()

shots = 10000


actual_theta = np.pi / 3



qc = QuantumCircuit(1, 1)


qc.ry(actual_theta, 0)


qc.measure(0, 0)



compiled_circuit = transpile(qc, simulator)

job = simulator.run(
    compiled_circuit,
    shots=shots
)

result = job.result()

counts = result.get_counts()



count_1 = counts.get('1', 0)

P1 = count_1 / shots



estimated_theta = 2 * np.arcsin(
    np.sqrt(P1)
)


print("Measurement Counts:", counts)

print("\nMeasured P(1):", P1)

print("\nActual Angle:")
print(np.degrees(actual_theta), "degrees")

print("\nEstimated Angle:")
print(np.degrees(estimated_theta), "degrees")
```

```python
import numpy as np

from qiskit import QuantumCircuit, transpile
from qiskit_aer import AerSimulator

from scipy.stats import binomtest


simulator = AerSimulator()

shots = 10000




qc = QuantumCircuit(1, 1)


qc.h(0)


qc.measure(0, 0)


compiled_circuit = transpile(qc, simulator)

job = simulator.run(
    compiled_circuit,
    shots=shots
)

result = job.result()

quantum_counts = result.get_counts()

q0 = quantum_counts.get('0', 0)
q1 = quantum_counts.get('1', 0)


print("Quantum Random Generator:")
print(quantum_counts)

print("P(1) =", q1 / shots)



classical_bits = np.random.choice(
    [0, 1],
    size=shots,
    p=[0.4, 0.6]
)

c0 = np.sum(classical_bits == 0)
c1 = np.sum(classical_bits == 1)


print("\nBiased Classical Generator:")

print("Number of 0s:", c0)
print("Number of 1s:", c1)

print("P(1) =", c1 / shots)




quantum_test = binomtest(
    q1,
    shots,
    p=0.5,
    alternative='two-sided'
)

classical_test = binomtest(
    c1,
    shots,
    p=0.5,
    alternative='two-sided'
)


print("\n------------------------------")
print("STATISTICAL TEST RESULTS")
print("------------------------------")


print("\nQuantum Generator:")

print("p-value =", quantum_test.pvalue)

if quantum_test.pvalue > 0.05:
    print("Result: Cannot reject H0")
    print("Data is consistent with an unbiased 50/50 source.")
else:
    print("Result: Reject H0")
    print("Data significantly differs from a 50/50 source.")


print("\nBiased Classical Generator:")

print("p-value =", classical_test.pvalue)

if classical_test.pvalue > 0.05:
    print("Result: Cannot reject H0")
else:
    print("Result: Reject H0")
    print("The data shows statistically significant bias.")
```
