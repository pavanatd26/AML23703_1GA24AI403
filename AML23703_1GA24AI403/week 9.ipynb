```python
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator

qc = QuantumCircuit(2, 2)


qc.h(0)
qc.cx(0, 1)


qc.measure([0, 1], [0, 1])

print(qc)


simulator = AerSimulator()
result = simulator.run(qc, shots=1024).result()

counts = result.get_counts()

print("Measurement results:")
print(counts)
```

```python
#medium
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator

simulator = AerSimulator()

def bell_state(name):
    qc = QuantumCircuit(2, 2)

    if name == "Phi+":
        qc.h(0)
        qc.cx(0, 1)

    elif name == "Phi-":
        qc.x(0)
        qc.h(0)
        qc.cx(0, 1)

    elif name == "Psi+":
        qc.h(0)
        qc.cx(0, 1)
        qc.x(0)

    elif name == "Psi-":
        qc.x(0)
        qc.h(0)
        qc.cx(0, 1)
        qc.x(0)

    qc.measure([0, 1], [0, 1])

    result = simulator.run(qc, shots=1024).result()
    return result.get_counts()


states = ["Phi+", "Phi-", "Psi+", "Psi-"]

for state in states:
    print(state, ":", bell_state(state))
```

```python
#hard
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator

qc = QuantumCircuit(2, 2)


qc.h(0)
qc.cx(0, 1)


qc.h(0)
qc.h(1)


qc.measure([0, 1], [0, 1])

print(qc)

simulator = AerSimulator()
result = simulator.run(qc, shots=1024).result()

counts = result.get_counts()

print("Hadamard-basis measurement:")
print(counts)
```

```python
#challenge
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator

qc = QuantumCircuit(3, 3)


qc.h(0)
qc.cx(0, 1)
qc.cx(0, 2)


qc.measure([0, 1, 2], [0, 1, 2])

print(qc)

simulator = AerSimulator()
result = simulator.run(qc, shots=1024).result()

counts = result.get_counts()

print("GHZ state:")
print(counts)
```
