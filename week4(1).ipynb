```python
qc.measure(3, 0)
qc.measure(4, 1)

print("Full Adder Circuit:")
print(qc)
```

```python
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector

import numpy as np
```

```python
theta = np.pi / 6       # 30 degrees

qc = QuantumCircuit(2)
```

```python
qc.ry(2 * theta, 0)
qc.cx(0, 1)
```

```python
print("Quantum Circuit:")
print(qc)
```

```python
state = Statevector.from_instruction(qc)

print("\nStatevector:")
print(state)
```

```python
print("\nAmplitudes:")

for basis, amplitude in zip(["00", "01", "10", "11"], state.data):
    print(f"|{basis}> : {amplitude:.4f}")
```

```python
expected = np.array([
    np.cos(theta),   # |00>
    0,
    0,
    np.sin(theta)    # |11>
])

print("\nExpected amplitudes:")
print(expected)
```

```python
if np.allclose(state.data, expected):
    print("\nPASS: Statevector matches the desired entangled state.")
else:
    print("\nFAIL: Statevector does not match.")
```

```python
concurrence = 2 * abs(
    np.cos(theta) * np.sin(theta)
)

print(f"\nConcurrence: {concurrence:.4f}")

if concurrence > 0:
    print("PASS: The state is entangled.")
else:
    print("The state is not entangled.")
```
